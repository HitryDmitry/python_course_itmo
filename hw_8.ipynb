```python
import requests
import pandas as pd
```

**Работа с API**

1. Получить информацию обо всех книгах и сохранить в pd.DataFrame


```python
all_books = requests.get('https://www.anapioficeandfire.com/api/books')
print(all_books.json())
```

```python
books_df = pd.DataFrame(all_books.json())
books_df
```

2. Получить информацию обо всех домах Вестероса (Houses) и сохранить в pd.DataFrame

```python
all_houses = requests.get('https://www.anapioficeandfire.com/api/houses')
houses_df = pd.DataFrame(all_houses.json())
houses_df
```

3. Получить информацию обо всех домах Вестероса, у которых есть девиз, и сохранить в pd.DataFrame

```python
payload = {'hasTitles':'True'}
all_houses_has_titles = requests.get('https://www.anapioficeandfire.com/api/houses', params=payload)
houses_df = pd.DataFrame(all_houses_has_titles.json())
houses_df
```

```python
houses_df['titles'].value_counts()
```

Видим, что строк с пустым параметром titles нет.
